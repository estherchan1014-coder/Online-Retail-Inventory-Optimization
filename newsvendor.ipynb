```python
import pandas as pd
import numpy as np
from scipy.stats import norm
```

```python
sku_weekly = pd.read_csv("sku_22423_weekly_demand.csv")

sku_weekly.head()
```

```python
mean_demand = sku_weekly["Quantity"].mean()
std_demand = sku_weekly["Quantity"].std()

print(f"Mean Demand: {mean_demand:.2f}")
print(f"Standard Deviation: {std_demand:.2f}")
```

### Assumptions for Newsvendor Analysis
*The Online Retail dataset provides historical demand information but does not contain inventory cost parameters such as purchase cost, selling price, salvage value, or stockout cost. As a result, these parameters cannot be estimated directly from the dataset.*

*To demonstrate how the Newsvendor model supports inventory decision-making, three hypothetical business scenarios are developed. Each scenario represents a different inventory environment and reflects different trade-offs between the costs of overstocking and understocking.*

*The analysis is intended to illustrate how changes in inventory economics affect the optimal order quantity while using the observed demand distribution of SKU 22423*

### Scenario 1: Balanced Cost Structure

```python
selling_price = 10
purchase_cost = 6
salvage_value = 2

Cu = selling_price - purchase_cost
Co = purchase_cost - salvage_value

critical_ratio = Cu / (Cu + Co)

z = norm.ppf(critical_ratio)

Q = mean_demand + z * std_demand

print("Cu =", Cu)
print("Co =", Co)
print("Critical Ratio =", round(critical_ratio, 4))
print("Z-score =", round(z, 4))
print("Optimal Q =", round(Q))
```

### Scenario 2: Stockouts are Expensive

```python
selling_price = 15
purchase_cost = 6
salvage_value = 2

Cu = selling_price - purchase_cost
Co = purchase_cost - salvage_value

critical_ratio = Cu / (Cu + Co)

z = norm.ppf(critical_ratio)

Q_star = mean_demand + z * std_demand

print("Cu =", Cu)
print("Co =", Co)
print("Critical Ratio =", round(critical_ratio, 4))
print("Z-score =", round(z, 4))
print("Optimal Q =", round(Q_star))
```

### Scenario 3: Excess Inventory is Costly

```python
selling_price = 8
purchase_cost = 6
salvage_value = 2

Cu = selling_price - purchase_cost
Co = purchase_cost - salvage_value

critical_ratio = Cu / (Cu + Co)

z = norm.ppf(critical_ratio)

Q_star = mean_demand + z * std_demand

print("Cu =", Cu)
print("Co =", Co)
print("Critical Ratio =", round(critical_ratio, 4))
print("Z-score =", round(z, 4))
print("Optimal Q =", round(Q_star))
```

### Comparison Table

```python
results = pd.DataFrame({"Scenario": ["Balanced Cost Structure", "Stockouts are Expensive", "Excess Inventory is Costly"],
    "Optimal Q": [233, 288, 187]})

results
```

### Conclusion

*The recommended order quantity varies across scenarios because the Newsvendor model depends on the relative costs of understocking and overstocking.*

*When stockouts become more expensive, the model recommends a larger order quantity (288 units). When excess inventory becomes more costly, the recommended order quantity decreases to 187 units. The balanced scenario results in an intermediate order quantity of 233 units.*

*These results demonstrate how inventory decisions depend not only on demand uncertainty but also on the economic consequences of inventory-related risks.*
